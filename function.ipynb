```python
#Take the username and password as input and authenticate the user
user = "Admin"
passw = "Password123"

def authenticate_user(username, password):
    """This function authenticates the user based on the provided username and password."""
    # For demonstration purposes, let's assume we have a simple authentication mechanism
    if user == username and passw == password:
        return True
    else:
        return False
validation_result = authenticate_user("Admin", "Password123")
if validation_result:
    print("Authentication successful!")
else:
    print("Authentication failed. Please check your username and password.")
```

```python
#Arguments types
def Emp_Registration(EmpId, EmpName, EmpDept):
    """This function registers an employee with the given details."""
    print(f"Employee ID: {EmpId}, Employee Name: {EmpName}, Employee Department: {EmpDept}")

Emp_Registration(101, "Rahul Agrawal", "HR")
Emp_Registration(EmpId=102, EmpDept="Finance", EmpName="Priya Singh") #keyword arguments
```

```python
#Default Arguments
def Emp_Registration(EmpId, EmpName, EmpDept="General"):
    """This function registers an employee with the given details."""
    print(f"Employee ID: {EmpId}, Employee Name: {EmpName}, Employee Department: {EmpDept}")

Emp_Registration(103, "Ankit Sharma") #Department will be General by default
Emp_Registration(104, "Sonia Gupta", "IT") #Department will be IT
```

```python
#Non-default arguments cannot follow default arguments
def Emp_Registration(EmpId, EmpName, EmpDept="General", EmpSalary =0):
    """This function registers an employee with the given details."""
    print(f"Employee ID: {EmpId}, Employee Name: {EmpName}, Employee Department: {EmpDept}, Employee Salary: {EmpSalary}")

Emp_Registration(105, "Rohit Verma") # This will raise a SyntaxError because non-default argument EmpSalary cannot follow default argument EmpDept.
```

```python
#var-args in python
def Sum_fun(*args):
    """This function takes a variable number of arguments and returns their sum."""
    total = sum(args)
    print(type(args)) #<class 'tuple'>
    print(args) #(1, 2, 3)
    print(f"Arguments received: {args}") #Arguments received: (1, 2, 3)
    print(f"the sum of {args} is: {total}")
    return total

Sum_fun()
Sum_fun(1,2,3) #the sum of (1, 2, 3) is: 6
Sum_fun(1)
```
